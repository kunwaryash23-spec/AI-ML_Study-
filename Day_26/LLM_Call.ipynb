```python
from openai import OpenAI
import sys
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
import fitz
from sklearn.decomposition import TruncatedSVD
from sklearn.preprocessing import Normalizer
from sklearn.pipeline import make_pipeline
```

```python
import pypdf

pdf_path = (r"D:\Ai engineering course\Day_25\Micro_Syallabus.pdf")

doc = fitz.open(pdf_path)
text = ""
for page in doc :
    text += page.get_text() + "\n"


chunk_size = 300
overlap = 30

chunks = []

start = 0
while start<len(text):
    end = start + chunk_size
    chunks.append(text[start:end])
    start += chunk_size - overlap

for i, chunk in enumerate(chunks):
    print(f"\n------Chunk {i}------")
    print(chunk)
```

```python
embedder = make_pipeline(TfidfVectorizer(stop_words = "english"),
                            TruncatedSVD (n_components = 20, random_state = 0),
                             Normalizer ())
index = embedder.fit_transform(chunks)

def retrieve(question, k=3):
    q = embedder.transform([question])[0]
    scores = index @ q
    return [(chunks[i]) for i in np.argsort(-scores)[:k]]
```

```python
import os



client = OpenAI(
    api_key=os.environ["GEMINI_API_KEY"],   # set in .env / shell — never hardcode
    
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/"
)

system_prompt = f"""

You are an expert faq bot. You will review the user's question about certain syllabus topic and its related information.
do not give response in markdown format


I will give you context : {context}
Your output response should be:


Answer in a polite and understandable method so that the user can understand what they are getting in the course and syllabus and also in the end 
promote saarathi academy.

do not answer any question unrelated to our course and simply reject them.



"""
user_query = """

whats 2+2?

"""

response = client.chat.completions.create(
    model="gemini-3.5-flash-lite",
    messages=[
        {
            "role": "system",
            "content": system_prompt
        },

        {
            "role": "user",
            "content": user_query
        }
    ]
)
```
