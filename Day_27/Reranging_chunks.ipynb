```python
import fitz

pdf_path = r"D:\Ai engineering course\Day_27\economic_survey.pdf"

reader = fitz.open(pdf_path)

text = ""

for page in reader:
    text += page.get_text() + "\n"

# Basic chunking
chunk_size = 500

chunks = []

for i in range(0, len(text), chunk_size):
    chunk_text = text[i:i + chunk_size]

    chunks.append({
        "chunk_id": len(chunks),
        "text": chunk_text
    })

# Display chunks
for chunk in chunks:
    print("Chunk ID:", chunk["chunk_id"])
    print(chunk["text"])
    print("-" * 50)
```
