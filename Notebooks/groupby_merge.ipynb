```python
"""

merge student.csv and scores.csv 
average score per subject with groupby
average score per city(needs the merge first)
which students score >= 90 in any subject


"""
```

```python
import numpy as np
import pandas as pd
```

```python
students = pd.read_csv(r"D:\Ai engineering course\Notebooks\students.csv")
```

```python
students
```

```python
scores = pd.read_csv(r"D:\Ai engineering course\Notebooks\scores.csv")
```

```python
scores
```

```python
ggg = pd.merge(students, scores, on = "id")
```

```python
ggg
```

```python
ggg.groupby("subject")["score"].mean()
```

```python
ggg.groupby("city")["score"].mean()
```

```python
high_scores = ggg[ggg['score'] >= 90]
```

```python
high_scores
```

```python
ggg.groupby('city').agg({'score': 'mean', 'name': 'count'})
```

```python
ggg.sort_values( by = 'score', ascending = False)
```

```python
ggg.to_csv("city.csv" , index = False)
```
