```python
import os
from datasets import Dataset
from ragas import evaluate

# 1. Use the Modern RAGAS Classes
from ragas.metrics import (
    ContextPrecision,
    ContextRecall,
    Faithfulness,
    AnswerRelevancy
)

from langchain_google_genai import ChatGoogleGenerativeAI, GoogleGenerativeAIEmbeddings

# 2. Setup your Gemini Key
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")  # key read from .env, never hardcoded

# 3. Initialize the Native Google Models
gemini_llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash")
gemini_embeddings = GoogleGenerativeAIEmbeddings(model="models/text-embedding-004")

# 4. Use the NEW RAGAS v0.2+ Schema Names!
data_samples = {
    "user_input": ["What does the learning rate control?"],
    "retrieved_contexts": [["The learning rate controls how big of a step the model takes during training. It is a crucial hyperparameter."]],
    "response": ["The learning rate controls how big of a step the model takes."],
    "reference": ["It controls the step size during model training."]
}
dataset = Dataset.from_dict(data_samples)

# 5. Evaluate
result = evaluate(
    dataset=dataset,
    metrics=[
        ContextPrecision(),
        ContextRecall(),
        Faithfulness(),
        AnswerRelevancy(),
    ],
    llm=gemini_llm, 
    embeddings=gemini_embeddings 
)

print(result)
```
