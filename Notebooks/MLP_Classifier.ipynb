```python
models = { 
"random ": make_pipeline(StandardScaler(), RandomForestClassifier (max_depth = 10 , random_state = 42)),
"neighbor" : make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=5)),
"decision" : make_pipeline(StandardScaler(), DecisionTreeClassifier (max_depth = 5 , random_state = 42)),
"logistic" : make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000)),
"net" : make_pipeline(StandardScaler(),MLPClassifier(hidden_layer_sizes = (16, 16), max_iter = 2000, random_state= 42)),
"poly_logistic": make_pipeline(PolynomialFeatures(degree=2, include_bias=False),StandardScaler(), LogisticRegression()
)

}
for name, pipeline in models.items():
    mean_score = cross_val_score(pipeline, X, healthy, cv=5).mean()
    print(f"{name}: {mean_score:}")
```
