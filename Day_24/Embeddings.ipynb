```python
import numpy as np
```

```python

# each word = 3 numbers we chose: [ royalty, male-ness, female-ness ]
words = {
    "king": np.array([0.9, 0.9, 0.1]),
    "queen": np.array([0.9, 0.1, 0.9]),
    "man":np.array([0.1, 0.9, 0.1]),
    "woman":np.array([0.1, 0.1, 0.9]),
}
def cosine(a, b):
    # 1 = same direction(same meaning),0 = neutral, -1 = comepletely different direction(opposite meaning)
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

print(cosine(words["king"], words["queen"]))
print(cosine(words["king"],words["woman"]))
```
